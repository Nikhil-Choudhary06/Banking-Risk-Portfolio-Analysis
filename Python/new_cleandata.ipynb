```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:/Users/Shubham/Downloads/PowerBI_Banking_Raw_Transactions (1).csv")
```

```python
df['Transaction_Date'] = pd.to_datetime(df['Transaction_Date'],errors= "coerce")
```

```python
df['Year']= pd.to_datetime(df['Transaction_Date']).dt.year
df['Month']= pd.to_datetime(df['Transaction_Date']).dt.month
df['Month_name']= pd.to_datetime(df['Transaction_Date']).dt.strftime('%b')
df['Year_Month']= pd.to_datetime(df['Transaction_Date']).dt.strftime('%Y-%m')
df['Day_Name']= pd.to_datetime(df['Transaction_Date']).dt.day_name()
```

```python
df
```

```python
df.columns
```

```python
txt_cols = ['Branch','Region','Branch_Segment','Customer_Segment','Product', 'Transaction_Type', 'Channel','Status']
for i in txt_cols:
    df[i]= df[i].astype(str).str.strip()
```

```python
df['Amount_INR']= df['Amount_INR'].round(2)
```

```python
df = df.drop_duplicates(subset= ['Transaction_ID'])
```

```python
df['DPD_Bucket']= df['DPD_Bucket'].str.strip()
```

```python
df['DPD_Bucket']= df['DPD_Bucket'].fillna('Current')
```
